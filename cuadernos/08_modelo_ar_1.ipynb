El modelo AR(1) (Autoregressive Model of Order 1, o Modelo Autorregresivo de primer orden) es uno de los modelos más simples y utilizados en el análisis de series de tiempo. Su principio fundamental es que el valor actual de una variable depende linealmente de su valor inmediatamente anterior más un término aleatorio.

### Definición matemática

Un proceso AR(1) se expresa como:

$X_t = \phi X_{t-1} + \varepsilon_t$

donde:
- $X_t$: valor de la serie en el instante $t$.
- $X_{t-1}$: valor de la serie en el instante anterior.
- $\phi$: coeficiente autorregresivo.
- $\varepsilon_t$: ruido blanco con media cero y varianza constante $\sigma^2$.

Cuando la serie tiene una media distinta de cero, suele escribirse como:

$$X_t = c + \phi X_{t-1} + \varepsilon_t$$

donde $c$ es una constante.

Interpretación física
El modelo AR(1) establece que el estado presente conserva una "memoria" del pasado inmediato.

Por ejemplo:
- Temperatura ambiental actual.
- Humedad relativa actual.
- Concentración de PM2.5.
- Velocidad del viento.
Normalmente estas variables no cambian bruscamente; el valor actual se parece mucho al valor observado hace unos minutos u horas.

Si la velocidad del viento fue de 5 m/s hace 10 minutos, es probable que el valor actual siga cercano a 5 m/s.

### Propiedad de estacionariedad

Para que un proceso AR(1) sea estacionario debe cumplirse:

$$|\phi| < 1$$

Esto significa que la influencia de los valores pasados disminuye conforme transcurre el tiempo.

**Caso 1:** $0 < \phi < 1$

Existe persistencia positiva.

Ejemplo:

$$\phi = 0.8$$

Si la serie aumenta en un instante, tenderá a permanecer elevada durante varios periodos.

**Caso 2:** $-1 < \phi < 0$

Existe correlación negativa.

Ejemplo:

$$\phi = -0.8$$

Los valores tienden a oscilar:
- valor alto
- valor bajo
- valor alto
- valor bajo

**Caso 3:** $\phi = 1$

Se obtiene un paseo aleatorio:

$$X_t = X_{t-1} + \varepsilon_t$$

La serie deja de ser estacionaria.

**Caso 4:** $|\phi| > 1$

El proceso es explosivo.

Pequeñas perturbaciones crecen indefinidamente.

Media del proceso

Para el modelo

$$X_t=c+\phi X_{t-1}+\varepsilon_t$$

la media estacionaria es:

$$\mu=\frac{c}{1-\phi}$$

Varianza del proceso

La varianza estacionaria es:

$$Var(X_t)=\frac{\sigma_\varepsilon^2}{1-\phi^2}$$

donde:
- $\sigma_\varepsilon^2$ es la varianza del ruido blanco.

Puede observarse que conforme $\phi$ se acerca a 1, la varianza aumenta considerablemente.

Función de autocorrelación (ACF)

Una característica importante del AR(1) es que su autocorrelación decrece exponencialmente:

$$\rho(k)=\phi^k$$

donde:
- $k$ es el retardo (lag).

Por ejemplo, si:

$$\phi=0.8$$

entonces:

| Retardo | Autocorrelación |
| :---: | :---: |
| 1 | 0.80 |
| 2 | 0.64 |
| 3 | 0.512 |
| 4 | 0.410 |

La memoria del sistema disminuye gradualmente.

Representación mediante operador de retardo

Definiendo el operador:

$$BX_t=X_{t-1}$$

el modelo puede escribirse como:

$$(1-\phi B)X_t=\varepsilon_t$$

Esta representación es muy útil para desarrollar la teoría de los modelos ARIMA.

Predicción con AR(1)

Si se conoce el valor actual $X_t$, la predicción a un paso adelante es:

$$\hat X_{t+1}=\phi X_t$$

La predicción a dos pasos es:

$$\hat X_{t+2}=\phi^2 X_t$$

En general:

$$\hat X_{t+h}=\phi^h X_t$$

A medida que aumenta el horizonte $h$, la predicción converge hacia la media del proceso.

### Ejemplo numérico

Suponga el modelo:

$$X_t=0.7X_{t-1}+\varepsilon_t$$

y que:

$$X_{10}=20$$

Entonces:

**Predicción a un paso:**

$$\hat X_{11}=0.7(20)=14$$

**Predicción a dos pasos:**

$$\hat X_{12}=0.7(14)=9.8$$

**Predicción a tres pasos:**

$$\hat X_{13}=0.7(9.8)=6.86$$

La influencia del dato inicial disminuye progresivamente.

el AR(1) suele emplearse para:
- Predicción de temperatura en invernaderos.
- Predicción de humedad relativa.
- Modelado de déficit de presión de vapor (DPV).
- Predicción de velocidad del viento.
- Predicción de concentración de contaminantes (PM2.5 y PM10).
- Modelado de radiación solar.
Sin embargo, para variables con comportamientos más complejos suele extenderse a modelos:
- AR(p)
- ARMA
- ARIMA
- SARIMA
- VAR
- Redes neuronales recurrentes (LSTM, GRU)
- Modelos híbridos Grey–Deep Learning
que capturan dependencias de mayor longitud y patrones no lineales.


Resumen
El modelo AR(1) supone que el valor actual de una serie temporal depende únicamente del valor inmediatamente anterior y de una perturbación aleatoria. Es el bloque fundamental de los modelos autorregresivos y constituye la base teórica de gran parte del análisis moderno de series de tiempo, identificación de sistemas y predicción de variables ambientales.

### 1. Se plantea el modelo AR(1)
El modelo general es:

$$T_{a,t}=c+\phi T_{a,t-1}+\varepsilon_t$$

donde $T_{a,t}$ es el valor actual y $T_{a,t-1}$ es el valor inmediatamente anterior.

### 2. Se construyen dos columnas
A partir de la serie original se forman pares:

$$X=T_{a,t-1}$$

$$Y=T_{a,t}$$

Ejemplo:

| $T_{a,t-1}$ | $T_{a,t}$ |
| :---: | :---: |
| 28.45 | 28.03 |
| 28.03 | 28.03 |
| 28.03 | 28.79 |
| 28.79 | 28.37 |

Así, el modelo AR(1) se estima como una regresión lineal:

$$Y=c+\phi X$$

### 3. Se calculan las medias
Se obtiene la media de $X$ y de $Y$:

$$\bar X=22.8308$$

$$\bar Y=22.8357$$

### 4. Se calcula el coeficiente autorregresivo $\phi$
Se usa la fórmula de mínimos cuadrados:

$$\phi=\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{\sum (X_i-\bar X)^2}$$

Con los datos:

$$\sum (X_i-\bar X)(Y_i-\bar Y)=22557.0994$$

$$\sum (X_i-\bar X)^2=22636.0075$$

Entonces:

$$\phi=\frac{22557.0994}{22636.0075}$$

$$\phi=0.9965$$

### 5. Se calcula el intercepto $c$

$$c=\bar Y-\phi\bar X$$

Sustituyendo:

$$c=22.8357-(0.9965)(22.8308)$$

$$c=0.0845$$

### 6. Modelo AR(1) obtenido
Por tanto, el modelo estimado es:

$$T_{a,t}=0.0845+0.9965T_{a,t-1}+\varepsilon_t$$

### 7. Predicción con el modelo
El último dato de la serie es:

$$T_{a,t-1}=34.17$$

Entonces:

$$\hat T_{a,t}=0.0845+0.9965(34.17)$$

$$\hat T_{a,t}=34.1354$$

### 8. Interpretación
El valor:

$$\phi=0.9965$$

indica que la temperatura tiene alta persistencia temporal. Es decir, el valor actual depende casi totalmente del valor anterior.

Como:

$$|\phi|<1$$

el modelo es estacionario teóricamente, aunque está muy cerca de una dinámica tipo caminata aleatoria.

El modelo final queda:

$$\hat T_{a,t}=0.0845+0.9965T_{a,t-1}$$

```python
# ============================================================
# CONSTRUCCIÓN DE PARES ORDENADOS PARA EL MODELO AR(1)
# Variable: Temperatura T
# Entorno: Google Colab
# ============================================================

import pandas as pd
from IPython.display import display

# ------------------------------------------------------------
# 1. CARGAR ARCHIVO CSV
# ------------------------------------------------------------
print("Seleccione el archivo CSV con la serie T:")

# Detección automática del separador
df = pd.read_csv("../data/datos_clima.csv", sep=None, engine="python",
                 encoding="utf-8-sig")

print("\nColumnas disponibles:")
print(df.columns.tolist())

display(df.head())

# ------------------------------------------------------------
# 2. SELECCIONAR LA VARIABLE DE TEMPERATURA
# ------------------------------------------------------------

# Cambiado de 'Ta' a 'T' para coincidir con tu archivo CSV
columna_temperatura = "T"

if columna_temperatura not in df.columns:
    raise ValueError(
        f"No existe la columna '{columna_temperatura}'. "
        f"Columnas disponibles: {df.columns.tolist()}"
    )

Ta = pd.to_numeric(
    df[columna_temperatura], errors="coerce"
)

if Ta.isna().any():
    raise ValueError(
        "La serie contiene valores faltantes o no numéricos. "
        "Revise los datos antes de construir los pares."
    )

if len(Ta) < 2:
    raise ValueError("Se requieren al menos dos observaciones.")

# ------------------------------------------------------------
# 3. CONSTRUIR PARES ORDENADOS AR(1)
# ------------------------------------------------------------

tabla_ar1 = pd.DataFrame({
    "t": range(2, len(Ta) + 1),
    "Ta_t-1": Ta.shift(1).iloc[1:].to_numpy(),
    "Ta_t": Ta.iloc[1:].to_numpy()
})

# ------------------------------------------------------------
# 4. MOSTRAR RESULTADOS
# ------------------------------------------------------------

print("\nNúmero de observaciones:", len(Ta))
print("Número de pares AR(1):", len(tabla_ar1))

print("\nPrimeros 10 pares:")
display(tabla_ar1.head(10))

print("\nÚltimos 10 pares:")
display(tabla_ar1.tail(10))

# ------------------------------------------------------------
# 5. EXPORTAR A EXCEL
# ------------------------------------------------------------

archivo_salida = "Tabla_Pares_AR1.xlsx"

serie_original = pd.DataFrame({
    "t": range(1, len(Ta) + 1),
    "Ta": Ta.to_numpy()
})

with pd.ExcelWriter(archivo_salida, engine="openpyxl") as writer:

    serie_original.to_excel(
        writer, sheet_name="Serie Original", index=False
    )

    tabla_ar1.to_excel(
        writer, sheet_name="Pares AR1", index=False
    )

print("\nArchivo generado:", archivo_salida)
```

```python

# ============================================================
# ESTIMACIÓN DEL MODELO AUTORREGRESIVO AR(1)
# MÉTODO: MÍNIMOS CUADRADOS ORDINARIOS
# VARIABLE: TEMPERATURA Ta
# ENTORNO: GOOGLE COLAB
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ============================================================
# 1. CARGAR ARCHIVO
# ============================================================

print("Seleccione el archivo CSV o Excel:")
df = pd.read_excel(
    "Tabla_Pares_AR1.xlsx",
)

df.columns = df.columns.str.strip()

print("\nColumnas encontradas:")
print(df.columns.tolist())

display(df.head())

# ============================================================
# 2. CONSTRUIR LOS PARES ORDENADOS
# ============================================================

# Seleccionar la columna que contiene la temperatura
columna = "Ta"

if columna not in df.columns:
    raise ValueError(
        f"No se encontró la columna '{columna}'. "
        f"Columnas disponibles: {df.columns.tolist()}"
    )

Ta = pd.to_numeric(df[columna], errors="coerce")

if Ta.isna().any():
    raise ValueError(
        "La serie contiene datos faltantes o no numéricos."
    )

if len(Ta) < 3:
    raise ValueError("La serie debe contener al menos 3 datos.")

X = Ta.iloc[:-1].to_numpy(dtype=float)
Y = Ta.iloc[1:].to_numpy(dtype=float)

n = len(X)

tabla = pd.DataFrame({
    "t": np.arange(2, len(Ta) + 1),
    "Ta_t-1": X,
    "Ta_t": Y
})

print("\nNúmero de observaciones originales:", len(Ta))
print("Número de pares ordenados:", n)

# ============================================================
# 3. CÁLCULO DE LAS MEDIAS
# ============================================================

X_media = np.mean(X)
Y_media = np.mean(Y)

# ============================================================
# 4. DESVIACIONES RESPECTO A LAS MEDIAS
# ============================================================

dx = X - X_media
dy = Y - Y_media

# ============================================================
# 5. SUMATORIAS Sxx Y Sxy
# ============================================================

Sxx = np.sum(dx**2)
Sxy = np.sum(dx * dy)

if np.isclose(Sxx, 0):
    raise ValueError(
        "Sxx es cero: no se puede estimar phi."
    )

# ============================================================
# 6. COEFICIENTE AUTORREGRESIVO
# ============================================================

phi = Sxy / Sxx

# ============================================================
# 7. INTERCEPTO
# ============================================================

c = Y_media - phi * X_media

# ============================================================
# 8. VALORES ESTIMADOS
# ============================================================

Y_estimado = c + phi * X

# ============================================================
# 9. RESIDUALES
# ============================================================

residuales = Y - Y_estimado

# ============================================================
# 10. MÉTRICAS DE DESEMPEÑO
# ============================================================

SSE = np.sum(residuales**2)
SST = np.sum((Y - Y_media)**2)

RMSE = np.sqrt(np.mean(residuales**2))
MAE = np.mean(np.abs(residuales))

R2 = 1 - SSE / SST if SST > 0 else np.nan

# ============================================================
# 11. TABLA DE CÁLCULOS INTERMEDIOS
# ============================================================

tabla["X-Xmedia"] = dx
tabla["Y-Ymedia"] = dy
tabla["(X-Xmedia)^2"] = dx**2
tabla["Producto_cruzado"] = dx * dy
tabla["Ta_estimado"] = Y_estimado
tabla["Residual"] = residuales
tabla["Residual^2"] = residuales**2

# ============================================================
# 12. PRESENTAR RESULTADOS
# ============================================================

print("\n" + "="*55)
print("RESULTADOS DEL MODELO AR(1)")
print("="*55)

print(f"Número de pares = {n}")
print(f"Media X          = {X_media:.8f}")
print(f"Media Y          = {Y_media:.8f}")
print(f"Sxx              = {Sxx:.8f}")
print(f"Sxy              = {Sxy:.8f}")
print(f"Phi              = {phi:.8f}")
print(f"Intercepto       = {c:.8f}")
print(f"RMSE             = {RMSE:.8f}")
print(f"MAE              = {MAE:.8f}")
print(f"R2               = {R2:.8f}")

print("\nModelo estimado:")
print(f"Ta(t) = {c:.8f} + ({phi:.8f})*Ta(t-1)")

if abs(phi) < 1:
    print("\nAR(1) estimado: condición |phi| < 1 satisfecha.")
else:
    print("\nAdvertencia: |phi| >= 1.")

print("\nPrimeras filas de la tabla:")
display(tabla.head(10))

# ============================================================
# 13. GRÁFICA: OBSERVADO VS ESTIMADO
# ============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    tabla["t"], Y,
    label="Temperatura observada",
    linewidth=1.3
)

plt.plot(
    tabla["t"], Y_estimado,
    label="Temperatura estimada AR(1)",
    linewidth=1.2,
    linestyle="--"
)

plt.xlabel("Índice temporal")
plt.ylabel("Temperatura (°C)")
plt.title("Modelo AR(1): Observado vs estimado")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# 14. GRÁFICA DE REGRESIÓN
# ============================================================

plt.figure(figsize=(7, 6))

plt.scatter(
    X, Y,
    alpha=0.5,
    s=15,
    label="Datos observados"
)

orden = np.argsort(X)

plt.plot(
    X[orden],
    Y_estimado[orden],
    linewidth=2,
    label="Regresión AR(1)"
)

plt.xlabel("Ta(t-1) [°C]")
plt.ylabel("Ta(t) [°C]")
plt.title("Identificación del modelo AR(1)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# 15. GRÁFICA DE RESIDUALES
# ============================================================

plt.figure(figsize=(13, 4))

plt.plot(
    tabla["t"],
    residuales,
    linewidth=1
)

plt.axhline(0, color="black", linestyle="--")

plt.xlabel("Índice temporal")
plt.ylabel("Residual (°C)")
plt.title("Residuales del modelo AR(1)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# 16. EXPORTAR RESULTADOS A EXCEL
# ============================================================

resumen = pd.DataFrame({
    "Parámetro": [
        "N original", "N pares",
        "Media X", "Media Y",
        "Sxx", "Sxy",
        "Phi", "Intercepto",
        "SSE", "SST",
        "RMSE", "MAE", "R2"
    ],
    "Valor": [
        len(Ta), n,
        X_media, Y_media,
        Sxx, Sxy,
        phi, c,
        SSE, SST,
        RMSE, MAE, R2
    ]
})

archivo_salida = "Resultados_Modelo_AR1.xlsx"

with pd.ExcelWriter(
    archivo_salida,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Serie Original",
        index=False
    )

    tabla.to_excel(
        writer,
        sheet_name="Calculos AR1",
        index=False
    )

    resumen.to_excel(
        writer,
        sheet_name="Resultados",
        index=False
    )

print("\nArchivo generado:", archivo_salida)
#files.download(archivo_salida)
```
