Modelo de medias móviles MA(q) en series de tiempo
El modelo de medias móviles (Moving Average, MA) es uno de los modelos fundamentales para el análisis y la predicción de series de tiempo. Junto con los modelos autorregresivos AR(p), constituye la base matemática de los modelos ARMA y ARIMA.

En el contexto del modelado de variables ambientales, como la temperatura y la humedad relativa, el modelo MA permite representar la dinámica de una serie temporal mediante la influencia de las perturbaciones aleatorias presentes y pasadas.

1. ¿Qué es un modelo MA?
Un modelo de medias móviles de orden $q$, denotado por MA(q), es un modelo estocástico en el que el valor actual de una variable depende de una combinación lineal del error aleatorio actual y de los errores correspondientes a los $q$ periodos anteriores.

Su expresión matemática general es:
$$\boxed{X_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\cdots+\theta_q\varepsilon_{t-q}}$$

Donde:

| Símbolo | Descripción |
| :---: | :--- |
| $X_t$ | Valor de la serie en el instante $t$ |
| $\mu$ | Media constante de la serie |
| $\varepsilon_t$ | Innovación o perturbación aleatoria actual |
| $\varepsilon_{t-i}$ | Innovación correspondiente a $i$ periodos anteriores |
| $\theta_i$ | Coeficiente de medias móviles asociado al rezago $i$ |
| $q$ | Orden del modelo MA |

Se supone habitualmente que las innovaciones constituyen un proceso de ruido blanco:
$$\varepsilon_t\sim WN(0,\sigma_\varepsilon^2)$$

Es decir:
$$E[\varepsilon_t]=0$$

$$Var(\varepsilon_t)=\sigma_\varepsilon^2$$

$$Cov(\varepsilon_t,\varepsilon_s)=0,\qquad t\neq s$$

Es importante distinguir que el ruido blanco no necesita seguir una distribución normal, aunque frecuentemente se supone normalidad para facilitar la inferencia estadística.

2. Interpretación del modelo MA
La idea fundamental es que el valor actual de una serie de tiempo está influenciado por perturbaciones que ocurrieron en periodos anteriores.

Consideremos la temperatura ambiental.

Supongamos que la temperatura fluctúa alrededor de un valor promedio de 25 °C. Estas fluctuaciones pueden estar asociadas con perturbaciones meteorológicas no explicadas explícitamente por el modelo.

Un modelo MA(2) podría expresar:
$$T_t=25+\varepsilon_t+0.6\varepsilon_{t-1}+0.3\varepsilon_{t-2}$$

Esto significa que la temperatura actual depende de:
- La temperatura media de 25 °C.
- La perturbación actual $\varepsilon_t$.
- El 60 % de la perturbación del periodo anterior, según el coeficiente del modelo.
- El 30 % de la perturbación de dos periodos anteriores.

Por ejemplo, si:
$$\varepsilon_t=0.4,\quad \varepsilon_{t-1}=0.8,\quad \varepsilon_{t-2}=-0.5$$

Entonces:
$$T_t=25+0.4+0.6(0.8)+0.3(-0.5)$$

$$\boxed{T_t=25.73\ ^\circ\text{C}}$$

El modelo representa una temperatura de 25.73 °C a partir de la media y de la combinación de perturbaciones.

Observación: los coeficientes 0.6 y 0.3 son parámetros del modelo, no porcentajes de la temperatura. Además, las innovaciones son cantidades latentes que se estiman a partir de la serie observada.

3. Diferencia entre MA y AR
Esta distinción es fundamental, especialmente si anteriormente hemos trabajado con modelos AR(1) y AR(2).

Un modelo AR(2) tiene la forma:
$$X_t=c+\phi_1X_{t-1}+\phi_2X_{t-2}+\varepsilon_t$$

Mientras que un modelo MA(2) se expresa como:
$$X_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}$$

La diferencia principal consiste en que el modelo AR utiliza valores anteriores de la propia variable, mientras que el MA utiliza perturbaciones aleatorias anteriores.

| **PACF teórica** | Se anula después del rezago $p$ | Decae gradualmente, en general |

4. Modelo MA(1)
El modelo más sencillo es el MA(1):
$$\boxed{X_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}}$$

Este modelo incorpora una perturbación anterior.

4.1 Media del proceso
Aplicando esperanza matemática:
$$E[X_t]=E[\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}]$$

Como las innovaciones tienen media cero:
$$\boxed{E[X_t]=\mu}$$

4.2 Varianza del proceso
La varianza es:
$$Var(X_t)=Var(\varepsilon_t+\theta_1\varepsilon_{t-1})$$

Debido a que las innovaciones están incorrelacionadas:
$$\boxed{Var(X_t)=\sigma_\varepsilon^2(1+\theta_1^2)}$$

4.3 Autocovarianza
La función de autocovarianza es:
$$\gamma(k)=Cov(X_t,X_{t-k})$$

Para un MA(1):
$$\gamma(0)=\sigma_\varepsilon^2(1+\theta_1^2)$$

$$\gamma(1)=\theta_1\sigma_\varepsilon^2$$

$$\gamma(k)=0,\qquad |k|>1$$

4.4 Autocorrelación
La función de autocorrelación se define como:
$$\rho(k)=\frac{\gamma(k)}{\gamma(0)}$$

Por lo tanto:
$$\boxed{\rho(1)=\frac{\theta_1}{1+\theta_1^2}}$$

Y:
$$\boxed{\rho(k)=0,\qquad |k|>1}$$

Esta es una propiedad fundamental: la autocorrelación teórica de un MA(1) desaparece después del primer rezago.

Por ejemplo, para $\theta_1=0.6$:
$$\rho(1)=\frac{0.6}{1+0.6^2}=0.4412$$

5. Modelo MA(2)
El modelo MA(2) incorpora dos innovaciones anteriores:
$$\boxed{X_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}}$$

Su media es:
$$E[X_t]=\mu$$

Su varianza:
$$\boxed{\gamma(0)=\sigma_\varepsilon^2(1+\theta_1^2+\theta_2^2)}$$

Las autocovarianzas son:
$$\gamma(1)=\sigma_\varepsilon^2\theta_1(1+\theta_2)$$

$$\gamma(2)=\sigma_\varepsilon^2\theta_2$$

$$\gamma(k)=0,\qquad |k|>2$$

Por consiguiente, las autocorrelaciones son:
$$\boxed{\rho(1)=\frac{\theta_1(1+\theta_2)}{1+\theta_1^2+\theta_2^2}}$$

$$\boxed{\rho(2)=\frac{\theta_2}{1+\theta_1^2+\theta_2^2}}$$

Para rezagos superiores a dos:
$$\rho(k)=0,\qquad |k|>2$$

Esto permite identificar un posible modelo MA(2) mediante la función de autocorrelación muestral.

6. Identificación del orden $q$
Una de las principales herramientas para identificar el orden de un modelo MA es la función de autocorrelación (ACF).

La regla teórica es:
- MA(1): la ACF se anula después del rezago 1.
- MA(2): la ACF se anula después del rezago 2.
- MA(q): la ACF se anula después del rezago $q$.
Sin embargo, cuando trabajamos con datos reales, las autocorrelaciones muestrales no son exactamente cero. Se utilizan intervalos de confianza y criterios de información como AIC y BIC para seleccionar modelos candidatos.

7. Estacionariedad e invertibilidad
Son dos propiedades importantes que deben distinguirse.

Estacionariedad: cualquier proceso MA(q) finito con coeficientes constantes e innovaciones de ruido blanco de varianza finita es estacionario en sentido débil.

Invertibilidad: permite expresar las innovaciones como una combinación convergente de las observaciones presentes y pasadas.

Para MA(1), con la convención de signos que estamos utilizando, la condición de invertibilidad es:
$$\boxed{|\theta_1|<1}$$

Por ejemplo, un MA(1) con $\theta_1=1.3$ sigue siendo estacionario, pero no es invertible.

Para MA(2), la invertibilidad exige que las raíces de:
$$1+\theta_1z+\theta_2z^2=0$$

estén fuera del círculo unitario, es decir, que tengan módulo mayor que uno.

8. Predicción mediante un modelo MA
Una característica interesante es que las predicciones de un MA(q) dependen de un número finito de innovaciones pasadas.

Para MA(2):
$$\quad X_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}$$

La predicción un paso adelante es:
$$\boxed{\hat X_{t+1|t}=\mu+\theta_1\hat\varepsilon_t+\theta_2\hat\varepsilon_{t-1}}$$

Para dos pasos:
$$\boxed{\hat X_{t+2|t}=\mu+\theta_2\hat\varepsilon_t}$$

Y para tres o más pasos:
$$\boxed{\hat X_{t+h|t}=\mu,\qquad h>2}$$

Esto ocurre porque las innovaciones futuras tienen esperanza condicional cero.

Por tanto, un MA(2) tiene una memoria predictiva limitada a dos pasos. A horizontes mayores, su predicción puntual converge inmediatamente a la media.

9. Diferencia entre el modelo MA y el promedio móvil convencional
Existe una confusión frecuente entre dos conceptos que comparten el nombre de medias móviles.

| Modelo MA(q) estocástico | Promedio móvil simple |
| :--- | :--- |
| Utiliza innovaciones actuales y anteriores | Utiliza observaciones anteriores |
| Es un modelo probabilístico | Es un procedimiento de suavizado |
| Estima parámetros $\theta_i$ | Calcula promedios sobre una ventana |
| Se utiliza para modelar y predecir | Se utiliza principalmente para suavizar tendencias y fluctuaciones |

Por ejemplo, un promedio móvil simple de tres observaciones es:
$$SMA_t=\frac{X_t+X_{t-1}+X_{t-2}}{3}$$

Esta ecuación no corresponde a un modelo estocástico MA(2).

10. Aplicación a una serie temporal de temperatura
Consideremos una estación meteorológica que registra temperatura cada cinco minutos.

Si se ajusta un modelo MA(2), su expresión podría ser:
$$T_t=\mu+\varepsilon_t+\theta_1\varepsilon_{t-1}+\theta_2\varepsilon_{t-2}$$

En este caso:
- $\varepsilon_{t-1}$ representa una innovación ocurrida cinco minutos antes.
- $\varepsilon_{t-2}$ representa una innovación ocurrida diez minutos antes.
- Los parámetros $\theta_1$ y $\theta_2$ cuantifican la influencia de esas innovaciones.

Para implementar correctamente el modelo, conviene seguir una secuencia metodológica: inspeccionar la serie, verificar su regularidad temporal, analizar estacionariedad, examinar ACF y PACF, estimar los parámetros, diagnosticar los residuales y evaluar predicciones fuera de muestra.

En series ambientales como la temperatura, que suelen presentar ciclos diarios y tendencias, puede ser necesario eliminar componentes deterministas o diferenciar la serie antes de ajustar un MA(q).

considerando la temperatura ($T$) como variable de interés y utilizando el modelo estocástico de medias móviles $MA(q)$.

El modelo matemático será:
$$\boxed{T_t=\mu+\varepsilon_t+\sum_{i=1}^{q}\theta_i\varepsilon_{t-i}}$$

```python

# ============================================================
# MODELADO DE SERIES DE TIEMPO - MA(q)
# Variable de interés: Temperatura (T)
# Archivo: datos_clima(1).csv
# Entorno: Google Colab
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

from google.colab import files
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.stats.diagnostic import acorr_ljungbox
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. CARGA DEL DATASET
# ------------------------------------------------------------

print("Seleccione el archivo CSV:")
uploaded = files.upload()

archivo = next(iter(uploaded))
df = pd.read_csv(archivo, encoding="utf-8-sig")

print("\nDimensiones originales:", df.shape)
print("\nColumnas:", df.columns.tolist())
print(df.head())

# ------------------------------------------------------------
# 2. PREPARACIÓN DE LA SERIE
# ------------------------------------------------------------

df["FechaHora"] = pd.to_datetime(
    df["Fecha"].astype(str) + " " + df["Hora"].astype(str),
    format="%d/%m/%y %H:%M:%S",
    errors="coerce"
)

df["T"] = pd.to_numeric(df["T"], errors="coerce")

df = (
    df.dropna(subset=["FechaHora", "T"])
      .sort_values("FechaHora")
      .drop_duplicates(subset="FechaHora")
      .reset_index(drop=True)
)

# Detectar intervalos temporales
diferencias = df["FechaHora"].diff()

print("\nIntervalos más frecuentes:")
print(diferencias.value_counts().head(10))

# Identificar segmentos continuos de un minuto
salto = diferencias.ne(pd.Timedelta(minutes=1))
df["segmento"] = salto.cumsum()

segmentos = df.groupby("segmento").size()

print("\nSegmentos continuos:")
print(segmentos)

# Seleccionar el segmento continuo más largo
id_segmento = segmentos.idxmax()

serie = (
    df.loc[df["segmento"] == id_segmento,
           ["FechaHora", "T"]]
      .set_index("FechaHora")["T"]
      .asfreq("min")
)

print("\nRegistros utilizados:", len(serie))
print("Inicio:", serie.index.min())
print("Fin:", serie.index.max())

# ------------------------------------------------------------
# 3. VISUALIZACIÓN DE LA SERIE
# ------------------------------------------------------------

plt.figure(figsize=(14, 4))
plt.plot(serie.index, serie.values, linewidth=0.8)
plt.title("Serie temporal original de temperatura")
plt.xlabel("Fecha y hora")
plt.ylabel("Temperatura (°C)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. PARTICIÓN CRONOLÓGICA
# ------------------------------------------------------------

n = len(serie)
n_train = int(n * 0.80)

train = serie.iloc[:n_train]
test = serie.iloc[n_train:]

print("\nEntrenamiento:", len(train))
print("Prueba:", len(test))

# ------------------------------------------------------------
# 5. PRUEBA DE ESTACIONARIEDAD ADF
# ------------------------------------------------------------

resultado_adf = adfuller(
    train,
    autolag="AIC"
)

print("\nPRUEBA AUGMENTED DICKEY-FULLER")
print("Estadístico ADF:", resultado_adf[0])
print("p-valor:", resultado_adf[1])

if resultado_adf[1] < 0.05:
    print("Se rechaza la hipótesis de raíz unitaria.")
    print("La serie podría ser estacionaria.")
else:
    print("No se rechaza la hipótesis de raíz unitaria.")
    print("Se recomienda examinar diferenciación o tendencia.")

# ------------------------------------------------------------
# 6. ACF Y PACF DE ENTRENAMIENTO
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plot_acf(train, lags=40, ax=axes[0])
plot_pacf(train, lags=40, ax=axes[1], method="ywm")

axes[0].set_title("ACF - Temperatura")
axes[1].set_title("PACF - Temperatura")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. SELECCIÓN DEL ORDEN MA(q)
# ------------------------------------------------------------

Q_MAX = 5

resultados = []
modelos = {}

for q in range(Q_MAX + 1):

    try:
        modelo = ARIMA(
            train,
            order=(0, 0, q),
            trend="c",
            enforce_stationarity=True,
            enforce_invertibility=True
        )

        ajuste = modelo.fit()

        modelos[q] = ajuste

        resultados.append({
            "q": q,
            "AIC": ajuste.aic,
            "BIC": ajuste.bic,
            "LogLik": ajuste.llf
        })

    except Exception as e:
        print(f"Error en MA({q}): {e}")

tabla = pd.DataFrame(resultados).sort_values("AIC")

print("\nCOMPARACIÓN DE MODELOS MA(q)")
print(tabla.to_string(index=False))

q_optimo = int(tabla.iloc[0]["q"])

print("\nOrden seleccionado:", q_optimo)

# ------------------------------------------------------------
# 8. ESTIMACIÓN DEL MODELO SELECCIONADO
# ------------------------------------------------------------

ajuste = modelos[q_optimo]

print("\nRESUMEN DEL MODELO")
print(ajuste.summary())

print("\nPARÁMETROS ESTIMADOS")
print(ajuste.params)

# ------------------------------------------------------------
# 9. PREDICCIÓN FUERA DE MUESTRA
# ------------------------------------------------------------

# Predicción dinámica desde el origen de entrenamiento.
# No utiliza valores observados del conjunto de prueba.

pred = ajuste.get_forecast(steps=len(test))

y_pred = pd.Series(
    np.asarray(pred.predicted_mean),
    index=test.index,
    name="Prediccion_MA"
)

# Intervalos de confianza del 95 %
ic = pred.conf_int(alpha=0.05)

lim_inf = np.asarray(ic.iloc[:, 0])
lim_sup = np.asarray(ic.iloc[:, 1])

# ------------------------------------------------------------
# 10. MÉTRICAS DE EVALUACIÓN
# ------------------------------------------------------------

y_real = test.to_numpy()
y_estimado = y_pred.to_numpy()

mae = mean_absolute_error(y_real, y_estimado)

rmse = np.sqrt(
    mean_squared_error(y_real, y_estimado)
)

r2 = r2_score(y_real, y_estimado)

print("\nMÉTRICAS EN CONJUNTO DE PRUEBA")
print(f"MAE  = {mae:.6f} °C")
print(f"RMSE = {rmse:.6f} °C")
print(f"R²   = {r2:.6f}")

# ------------------------------------------------------------
# 11. GRÁFICA DE PREDICCIÓN
# ------------------------------------------------------------

plt.figure(figsize=(14, 5))

plt.plot(
    test.index,
    y_real,
    label="Temperatura real",
    linewidth=1
)

plt.plot(
    test.index,
    y_estimado,
    label=f"Predicción MA({q_optimo})",
    linewidth=1.5
)

plt.fill_between(
    test.index,
    lim_inf,
    lim_sup,
    alpha=0.2,
    label="Intervalo de confianza 95 %"
)

plt.title("Predicción de temperatura mediante MA(q)")
plt.xlabel("Fecha y hora")
plt.ylabel("Temperatura (°C)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 12. ANÁLISIS DE RESIDUALES
# ------------------------------------------------------------

residuos = pd.Series(
    np.asarray(ajuste.resid),
    index=train.index
)

fig, axes = plt.subplots(2, 1, figsize=(13, 8))

axes[0].plot(residuos.index, residuos)
axes[0].axhline(0, color="red", linestyle="--")
axes[0].set_title("Residuales del modelo MA")
axes[0].set_ylabel("Error (°C)")
axes[0].grid(alpha=0.3)

plot_acf(
    residuos,
    lags=40,
    ax=axes[1]
)

axes[1].set_title("Autocorrelación de residuales")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 13. PRUEBA LJUNG-BOX
# ------------------------------------------------------------

lb = acorr_ljungbox(
    residuos,
    lags=[10, 20, 30],
    model_df=q_optimo,
    return_df=True
)

print("\nPRUEBA LJUNG-BOX")
print(lb)

print("\nUn p-valor pequeño indica evidencia")
print("de autocorrelación residual.")

# ------------------------------------------------------------
# 14. EXPORTACIÓN DE RESULTADOS
# ------------------------------------------------------------

salida = pd.DataFrame({
    "FechaHora": test.index,
    "Temperatura_real": y_real,
    "Temperatura_predicha": y_estimado,
    "Residual": y_real - y_estimado,
    "IC_inferior": lim_inf,
    "IC_superior": lim_sup
})

salida.to_csv(
    "resultados_modelo_MA.csv",
    index=False
)

tabla.to_csv(
    "comparacion_modelos_MA.csv",
    index=False
)

print("\nArchivos generados correctamente.")

files.download("resultados_modelo_MA.csv")
files.download("comparacion_modelos_MA.csv")
```

```python

# =====================================================
# VISUALIZACIÓN DE SERIES DE TIEMPO DESDE UN ARCHIVO CSV
# Entorno: Google Colab
# =====================================================

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# 1. Cargar archivo CSV
print("Seleccione el archivo CSV")
uploaded = files.upload()

nombre_archivo = next(iter(uploaded))

df = pd.read_csv(nombre_archivo, encoding="utf-8-sig")

# 2. Información general
print("\nDimensiones del dataset:")
print(df.shape)

print("\nColumnas disponibles:")
print(df.columns.tolist())

print("\nPrimeros registros:")
display(df.head())

# 3. Construir columna temporal, si existe
if {"Fecha", "Hora"}.issubset(df.columns):

    df["FechaHora"] = pd.to_datetime(
        df["Fecha"].astype(str) + " " +
        df["Hora"].astype(str),
        dayfirst=True,
        errors="coerce"
    )

elif "FechaHora" in df.columns:

    df["FechaHora"] = pd.to_datetime(
        df["FechaHora"],
        dayfirst=True,
        errors="coerce"
    )

# 4. Seleccionar eje horizontal
if "FechaHora" in df.columns:

    df = df.dropna(subset=["FechaHora"])
    df = df.sort_values("FechaHora")
    eje_x = df["FechaHora"]
    etiqueta_x = "Fecha y hora"

else:
    eje_x = df.index
    etiqueta_x = "Número de observación"

# 5. Identificar variables numéricas
excluir = ["FechaHora"]

for columna in df.columns:

    if columna not in excluir:
        convertido = pd.to_numeric(
            df[columna],
            errors="coerce"
        )

        if convertido.notna().sum() > 0:
            df[columna] = convertido

variables = df.select_dtypes(
    include="number"
).columns.tolist()

print("\nVariables numéricas:")
print(variables)

# 6. Graficar cada variable
for variable in variables:

    plt.figure(figsize=(14, 5))

    plt.plot(
        eje_x,
        df[variable],
        linewidth=1,
        label=variable
    )

    plt.title(f"Serie temporal de {variable}")
    plt.xlabel(etiqueta_x)
    plt.ylabel(variable)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

# 7. Gráfica conjunta de variables
plt.figure(figsize=(14, 5))

for variable in variables:

    plt.plot(
        eje_x,
        df[variable],
        linewidth=1,
        label=variable
    )

plt.title("Comparación de series temporales")
plt.xlabel(etiqueta_x)
plt.ylabel("Valores observados")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nVisualización finalizada.")
```
